```python
#Bubble Sort

def bubble_sort(list1):
  size = len(list1)

  for i in range(size-1):
    swaped = False  # took it because if the loop is already sorted then you dont have to iterate again. took this variable to keep track of it.
    for j in range(size-1-i):
      if list1[j] > list1[j+1]:
        temp = list1[j]
        list1[j] = list1[j+1]
        list1[j+1] = temp
        swaped = True

    if not swaped: #did not swaped any value, that means it already sorted. to broke the outer loop
      break


list1 = [2,5,1,0,10,6,3]
bubble_sort(list1)
print(list1)
```

```python
#selection sort

def selection_sort(arr):
  size = len(arr)

  for i in range(size-1):
    min_index = i
    for j in range(min_index+1, size): #min index ber korar jnno.
      if arr[j] < arr[min_index]:   #comparing values
        min_index = j

    if i != min_index: # min index jodi oi index er shoman na hoi
      arr[i], arr[min_index] = arr[min_index], arr[i]


list1 = [2,5,1,0,10,6,3]
selection_sort(list1)
print(list1)
```

```python
#insertion sort - o(n)

def insertion_sort(arr):
  for i in range(1, len(arr)):
    temp = arr[i]
    j = i-1

    while j >= 0 and temp < arr[j]:
      arr[j+1] = arr[j]
      j = j - 1

    arr[j+1] = temp

list1 = [-1, 2,5,1,0,10,6,3]
#for example i is right now in index 2. now we have to comapre it with the two variable before that.
#that is why we are taking another variable known as j.
# 2,5,5,0,10,6,3 --> value right e shift kore
# 2,2,5,0,10,6,3
# 1,2,5,0,10,6,3 --> loop theke ber hoye

insertion_sort(list1)
print(list1)
```

```python
from os import access
#Counting Sort

def CountingSort(arr):
  size = len(arr)
  max_value = max(arr)
  sorted_arr = [0]*size
  counting_list = [0]*(max_value+1) #normally max na diye 10 diye multiply korleo hoi. cause counting sort e range beshi hoi na.


  for i in range(0,size):
    counting_list[arr[i]] += 1

  for j in range(1, max_value+1):
    counting_list[j] = counting_list[j] + counting_list[j-1]

  access = size - 1

  while access >= 0:
        sorted_arr[counting_list[arr[access]] - 1] = arr[access]
        counting_list[arr[access]] -= 1
        access -= 1

  return sorted_arr


# list1 = [2,5,1,0,9,6,3]
list1 = [2,1,2,6,6,0,4,3,3]
print(CountingSort(list1))
```

```python
#counting sort with modification

def countsort(arr):
    small = arr[0]
    big = arr[0]

    for i in arr:
        if i < small:
            small = int(i)
        if i > big:
            big = int(i)

    new_arr = [None]*(big-small)

    for i in arr:
        if new_arr[int(i)] == None:
            new_arr[int(i)] = [i]
        else:
            new_arr[int(i)].append(i)


    return new_arr

arr =   [2, 5, 1.2, 6.7, 1.7, 9.3, 2.2, 7.7, 0, -4, -5.1, 2, 5, 5.2]

print(countsort(arr))
```

```python
# A Divide and Conquer based program
# for maximum subarray sum problem

# Find the maximum possible sum in
# arr[] auch that arr[m] is part of it


def maxCrossingSum(arr, l, m, h):

	# Include elements on left of mid.
	sm = 0
	left_sum = -10000

	for i in range(m, l-1, -1): #EKHANE LEFT SIDE ER MAX ASHE. MID THEKE ULTA DIK E JAITESE LEFT MAX
		sm = sm + arr[i]

		if (sm > left_sum):
			left_sum = sm

	# Include elements on right of mid
  #EKHANE RIGHT SIDE ER MAX ASHE
	sm = 0
	right_sum = -1000
	for i in range(m, h + 1):
		sm = sm + arr[i]

		if (sm > right_sum):
			right_sum = sm

	# Return sum of elements on left and right of mid
	# returning only left_sum + right_sum will fail for [-2, 1]
	return max(left_sum + right_sum - arr[m], left_sum, right_sum) #EKHANE MINUS KORTESI KAERON DUI JAIGAI E M NITESI. MINUS E KORLE EXTRA CHOLE JABE


# Returns sum of maximum sum subarray in aa[l..h]
def maxSubArraySum(arr, l, h):
	#Invalid Range: low is greater than high
	if (l > h):
		return -10000
	# Base Case: Only one element
	if (l == h):
		return arr[l]

	# Find middle point
	m = (l + h) // 2

	# Return maximum of following three possible cases
	# a) Maximum subarray sum in left half
	# b) Maximum subarray sum in right half
	# c) Maximum subarray sum such that the
	#	 subarray crosses the midpoint
	return max(maxSubArraySum(arr, l, m-1),
			maxSubArraySum(arr, m+1, h),
			maxCrossingSum(arr, l, m, h))


# Driver Code
arr = [6, -4, 5, 7, -13, 9]
n = len(arr)

max_sum = maxSubArraySum(arr, 0, n-1)
print("Maximum contiguous sum is ", max_sum)

# This code is contributed by Nikita Tiwari.
```
