```python
import numpy as np
```

```python
#Pass matrix
A = np.matrix([[0.25, 0.20, 0.1], [0.1, 0.25, 0.2],[0.1, 0.1, 0.25]])
#Goal vector
g = np.transpose(np.matrix([0.05, 0.15, 0.05]))
```

# Solution Using Linear Algebra

```python
xT1 = np.linalg.solve(np.identity(3) - A,g)

print('Expected Threat')
print('Central, Box, Wing')
print(np.transpose(xT1))
```

# Iterative Solution

```python
xT2=np.zeros((3,1))
for t in range(10):
   print(np.matmul(A,xT2) + g)
   xT2 = np.matmul(A,xT2) + g

print('Expected Threat')
print('Central, Box, Wing')
print(np.transpose(xT2))
```

# Monte Carlo Simulation

```python
num_sims=3
xT3=np.zeros(3)

description = {0: 'Central', 1: 'Wing', 2: 'Box' }

for i in range(3):
    num_goals = 0

    print('---------------')
    print('Start from ' + description[i] )
    print('---------------')

    for n in range(num_sims):

        ballinplay=True
        #Initial state is i
        s = i
        describe_possession=''

        while ballinplay:
            r=np.random.rand()

            # Make commentary text
            describe_possession = describe_possession + ' - ' + description[s]


            #Cumulative sum of in play probabilities
            c_sum=np.cumsum(A[s,:])
            new_s = np.sum(r>c_sum)
            if new_s>2:
                #Ball is either goal or out of play
                ballinplay=False
                if r < g[s] + c_sum[0,2]:
                    #Its a goal!
                    num_goals = num_goals + 1
                    describe_possession = describe_possession + ' - Goal!'
                else:
                    describe_possession = describe_possession + ' - Out of play'
            s = new_s

        print(describe_possession)

    xT3[i] = num_goals/num_sims


print('\n\n---------------')
print('Expected Threat')
print('Central, Box, Wing')
print(xT3)
```

**Expected Threat (xT)** for three pitch zones by simulating possessions.

Zones:
- `0 → Central`
- `1 → Wing`
- `2 → Box`


We simulate many possessions and estimate the **probability of scoring**.

`num_sims` = number of simulated possessions per zone  
`xT3` = stores xT values for each zone.

Example:  
If **1 goal occurs in 2 simulations**

xT = 1 / 2 = **0.5**

Each simulation starts from a zone.

The simulation then continues while the ball is in play.



At every step the code generates a random number:

`r = np.random.rand()`

Example:

`r = 0.63`

This random value determines what event happens next.



Matrix `A` contains the **passing probabilities between zones**.

Example (from Central):

Central → Central = 0.30  
Central → Wing = 0.40  
Central → Box = 0.20  

Total passing probability = **0.90**

Cumulative probabilities become:

`[0.30, 0.70, 0.90]`

Example:

If `r = 0.63`

0.30 < 0.63 < 0.70 → **ball moves to Wing**

Possession path becomes:

Central → Wing


If the random number is larger than the passing probability (here **> 0.90**), the possession ends.

Two possibilities:

Goal probability is stored in `g[s]`.

Example:

`g[Central] = 0.05`

Goal range:

0.90 → 0.95

If:

`r = 0.92` → **Goal**

If:

`r = 0.97` → **Out of play**

Example simulated possession:

Central → Wing → Box → Goal


After running the simulations:

`xT3[i] = num_goals / num_sims`

Example result:

Central: 1 goal in 10 simulations → xT = **0.10**  
Wing: 0 goals in 10 simulations → xT = **0.00**  
Box: 3 goals in 10 simulations → xT = **0.30**



Expected Threat measures:

xT(zone) = P(goal | possession starts in that zone)

By simulating many possessions, we estimate **how dangerous each area of the pitch is**.
