```python
import GEOparse
gse = GEOparse.get_GEO(geo="GSE6613", destdir="./")
```

```python
gse.phenotype_data
```

```python
samples = gse.pivot_samples('VALUE')
samples
```

```python
#Probe-to-Gene Mapping

gpl = gse.gpls[list(gse.gpls.keys())[0]]
print(gpl.table.columns)
probe_to_gene = dict(zip(gpl.table["ID"], gpl.table["Gene Symbol"]))
```

```python
# Transpose and matrix cleaning
X_geo = samples.T


X_geo = X_geo.apply(pd.to_numeric, errors='coerce')
X_geo = X_geo.dropna(axis=1)

print(f"Matrix shape (samples × genes): {X_geo.shape}")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_geo.values)
X_geo_scaled = scaler.transform(X_geo.values)

print(f"Scaled matrix shape: {X_geo_scaled.shape}")
print(f"Mean (should be ~0): {X_geo_scaled.mean():.4f}")
print(f"Std  (should be ~1): {X_geo_scaled.std():.4f}")
```

```python
#Running PCA

from sklearn.decomposition import PCA

pca_geo = PCA(n_components=0.95, random_state=2020)
pca_geo.fit(X_geo_scaled)
X_pca_geo = pca_geo.transform(X_geo_scaled)

print(f"Number of PCs selected (95% variance): {pca_geo.n_components_}")
print(f"Total variance explained: {sum(pca_geo.explained_variance_ratio_)*100:.2f}%")
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('ggplot')
plt.figure(figsize=(10, 5))
plt.plot(np.cumsum(pca_geo.explained_variance_ratio_ * 100))
plt.axhline(y=95, color='red', linestyle='--', label='95% threshold')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Variance Explained (%)')
plt.title('PCA on GSE6613 Gene Expression Data')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Loading for top 5 PCA

top5_loadings  = pca_geo.components_[:5]
top5_variance  = pca_geo.explained_variance_ratio_[:5]

print("Loadings shape:", top5_loadings.shape)
print("\nVariance explained by each of the top 5 PCs:")
for i, v in enumerate(top5_variance):
    print(f"  PC{i+1}: {v*100:.3f}%")
print(f"\n  Combined: {sum(top5_variance)*100:.2f}%")
```

```python
#Weighted Gene Score

weighted_5 = np.abs(top5_loadings) * top5_variance[:, np.newaxis]


gene_scores_5 = np.sum(weighted_5, axis=0)


print(f"Scores computed for {len(gene_scores_5)} genes")
print(f"Highest score: {gene_scores_5.max():.6f}")
print(f"Lowest  score: {gene_scores_5.min():.6f}")
```

```python
# Map Scores to Gene Names and Sort

probe_ids = list(X_geo.columns)


results_5 = pd.DataFrame({
    'Probe_ID':   probe_ids,
    'Score':      gene_scores_5,
    'Gene_Symbol': [probe_to_gene.get(p, 'Unknown') for p in probe_ids]
})


results_5 = results_5.sort_values('Score', ascending=False).reset_index(drop=True)
results_5.index = results_5.index + 1

results_5.head(20)
```

```python
#Print the Final Top 20 List (5 PCs)

top20_5pc = results_5.head(20)[['Probe_ID', 'Gene_Symbol', 'Score']]

print("=" * 100)
print("   TOP 20 GENES — USING TOP 5 PRINCIPAL COMPONENTS")
print("=" * 100)
print(f"{'Rank':<6} {'Probe ID':<15} {'Gene Symbol':<20} {'Score'}")
print("-" * 100)
for rank, row in top20_5pc.iterrows():
    print(f"{rank:<6} {row['Probe_ID']:<15} {str(row['Gene_Symbol']):<20} {row['Score']:.6f}")
```

```python
#  TOP 10 PCs
top10_loadings = pca_geo.components_[:10]
top10_variance = pca_geo.explained_variance_ratio_[:10]

print("Variance explained by each of the top 10 PCs:")
for i, v in enumerate(top10_variance):
    print(f"  PC{i+1}: {v*100:.3f}%")
print(f"\n  Combined: {sum(top10_variance)*100:.2f}%")

# Weighted scores
weighted_10    = np.abs(top10_loadings) * top10_variance[:, np.newaxis]
gene_scores_10 = np.sum(weighted_10, axis=0)


results_10 = pd.DataFrame({
    'Probe_ID':   probe_ids,
    'Score':      gene_scores_10,
    'Gene_Symbol': [probe_to_gene.get(p, 'Unknown') for p in probe_ids]
})

results_10 = results_10.sort_values('Score', ascending=False).reset_index(drop=True)
results_10.index = results_10.index + 1

# Print top 20
top20_10pc = results_10.head(20)[['Probe_ID', 'Gene_Symbol', 'Score']]

print("\n" + "=" * 55)
print("   TOP 20 GENES — USING TOP 10 PRINCIPAL COMPONENTS")
print("=" * 55)
print(f"{'Rank':<6} {'Probe ID':<15} {'Gene Symbol':<20} {'Score'}")
print("-" * 55)
for rank, row in top20_10pc.iterrows():
    print(f"{rank:<6} {row['Probe_ID']:<15} {str(row['Gene_Symbol']):<20} {row['Score']:.6f}")
```

```python
genes_5  = set(top20_5pc['Gene_Symbol'].tolist())
genes_10 = set(top20_10pc['Gene_Symbol'].tolist())

common      = genes_5 & genes_10
only_in_5   = genes_5 - genes_10
only_in_10  = genes_10 - genes_5

print(f"Genes in BOTH lists:       {len(common)}  → {common}")
print(f"Only in 5-PC list:         {len(only_in_5)}  → {only_in_5}")
print(f"Only in 10-PC list:        {len(only_in_10)}  → {only_in_10}")

# Side-by-side table
print("\n" + "=" * 100)
print(f"{'Rank':<6} {'Top 5 PCs':<22} {'Top 10 PCs'}")
print("-" * 100)
for i in range(20):
    g5  = str(top20_5pc.iloc[i]['Gene_Symbol'])
    g10 = str(top20_10pc.iloc[i]['Gene_Symbol'])
    marker = "✓" if g5 == g10 else " "
    print(f"{i+1:<6} {g5:<22} {g10}  {marker}")
```

##  Q/A

##   Top 5 PCs vs Top 10 PCs Gene Selection

### Q1: Is there any difference between the two lists?

Yes.

While some genes appear in both top-20 lists, several genes differ
between the 5-PC and 10-PC selections. The top-ranked genes tend to be
consistent (they dominate the most powerful PCs), but the lower-ranked
genes in the top-20 change as we include more PCs.

### Q2: Why is this happening?

Using 10 Principal Components (PCs) instead of 5 allows the analysis to capture subtle biological signals, such as disease-specific expression patterns, that are missed by the most dominant sources of variation. As these later PCs are integrated, genes that were previously "invisible" accumulate enough weight to enter the top-20 list. This process causes displacement within the rankings, particularly for genes at the bottom of the list where the influence of these subtler signals is most felt. Conversely, the top-ranked genes remain stable because their dominance is high enough to persist regardless of the number of components used.

### Q3: Is PCA a suitable approach for gene selection?

**Strengths:**
- Handles high-dimensional data well
- Completely unsupervised — no risk of overfitting to class labels
  during the selection step itself
- Captures global structure across all samples at once

**Weaknesses:**
- PCA is unsupervised, meaning it ignores class labels (Parkinson's vs.
  healthy). So the selected genes reflect the overall variation in the
  data, not necessarily the variation most relevant to the disease.
- Loadings are linear combinations of all genes — they're hard to
  interpret biologically compared to directly measured features.
- The number of PCs chosen (5 vs 10) significantly affects results,
  introducing a tuning decision with no universally correct answer.

**Conclusion:**
While PCA-based selection is an efficient tool for fast exploratory analysis, it lacks the specificity provided by class labels. For identifying disease biomarkers like those for Parkinson's, supervised methods—such as elastic net or supervised PCA—offer more biologically meaningful and reproducible results.
